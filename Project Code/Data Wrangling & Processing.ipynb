```python
'''
Individual .nc files are set to single years of cape and cin and then merged into a single dataset called "Data"

Once the datasets are merged, the output can be saved as a .zarr file using xarray's built-in .to_zarr functionality

'''
A = xr.open_dataset('cape.2018.nc') #Single year of daily cape
B = xr.open_dataset('cape.2019.nc')
C = xr.open_dataset('cape.2020.nc')
D = xr.open_dataset('cape.2021.nc')
a = xr.open_dataset('cin.2018.nc') #Single year of daily cin
b = xr.open_dataset('cin.2019.nc')
c = xr.open_dataset('cin.2020.nc')
d = xr.open_dataset('cin.2021.nc')

Data = xr.merge([A, B, C, D, a, b, c, d]) #merging individual .nc files into a single dataset
Data
```
